- Are there outliers?

Plots under distribution plot

- histplot
- kdeplot

```python
#  HERE FOR figure level -> displot


# axes level -> histplot -> kdeplot -> rugplot


# data distributed
# what tells - unviarte ragnge max min tells mean tells
# bimodel - multiple peaks
# outliers in the data
```

```python
# plotting univariate histogram
# histogram -- distribution  bins count

sns.histplot(data = tips,x ='total_bill')
```

```python
# displot kind = hist figure level
sns.displot(data=tips, x='total_bill', kind='hist')
```

```python
# bins parameter in histogram
sns.displot(data=tips, x='total_bill', kind='hist',bins=2)
```

```python
# It’s also possible to visualize the distribution of a categorical variable using the logic of a histogram.
# Discrete bins are automatically set for categorical variables
# ques - day wihc day how mnay customers come
# acting like a countplot this graph
# countplot
sns.displot(data=tips, x='day', kind='hist')
plt.plot()
```

```python
# hue parameter side by side two category
sns.displot(data=tips, x='tip', kind='hist',hue='sex')
```

```python
# element -> step clear
sns.displot(data=tips, x='tip', kind='hist',hue='sex',element='step')
```

# light color - small value

```python
# Heatmap work -------
# Plot rectangular data as a color-encoded matrix
# axes level function a heatmap
# only axes level function
# question - create a  grid information life exp every country
# lifexpentancy  value
# index - country name then columns year then value life exp

# increase size of the plot beacuse it is axes level function
plt.figure(figsize=(15,15))
gap_pivot=gap.pivot(index='country',columns='year',values='lifeExp')
sns.heatmap(gap_pivot)
```

```python
# annot customization
# only europeN Countries
temp_df = gap[gap['continent'] == 'Europe'].pivot(index='country',columns='year',values='lifeExp')
temp_df
plt.figure(figsize=(15,15))
#sns.heatmap(temp_df)
# question - no colour value need life expentency value  annot true
# annot with coliurs value also within heatmap
sns.heatmap(temp_df,annot=True)
```

```python
# linewidth
sns.heatmap(temp_df,annot=True,linewidth=0.1)
```

```python
# cmap - color maps color palette
sns.heatmap(temp_df,annot=True,cmap='viridis',linewidth=0.5)
```

```python
# correlation heatmap
# - parameter - corr - correlation matrix
# annot - correlation values inside cells
# 1.0 - perfect positive correlation +
cor=gap_pivot.corr(numeric_only=True)
plt.figure(figsize=(8,5))
sns.heatmap(cor,annot=True,cmap='coolwarm',fmt=".2f")
plt.show()
```

correlation used - to quickly identify strong positive and negative relationship among numerical features selection during eda
